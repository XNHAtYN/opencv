```python
import cv2

def im_show(name,picture):
    cv2.imshow("show",picture)
    cv2.waitKey()
    cv2.destroyAllWindows()
png=cv2.imread("picture.png")
im_show("show",png)
```

```python
cur_png=png.copy()
cur_png[0:500,0:500,0]=0
# cur_png[:,:,0]=0
cur_png[500:1000,500:1000,1]=0
im_show("r",cur_png)
```

```python
png_groy=cv2.imread("picture.png",cv2.IMREAD_GRAYSCALE)
im_show("show",png_groy)
```

```python
vc=cv2.VideoCapture("video.mp4")
if vc.isOpened():
    op=True
else:
    op=None
while op:
    rt,frame=vc.read()
    if frame is None:
        break
    if rt==True:
        gray=cv2.cvtColor(frame,cv2.COLOR_BGR2GRAY)
        cv2.imshow("result",gray)
        if cv2.waitKey(1) & 0xFF==27:
            break
vc.release()
cv2.destroyAllWindows()







```
